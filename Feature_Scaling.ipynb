```python
import pandas as pd
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.DataFrame({
    "Study_Hours": [2, 4, 6, 8],
    "Attendance": [50, 60, 70, 80]
})
```

```python
print(df)
```

```python
scaler = StandardScaler()
```

```python
scaled_data = scaler.fit_transform(df)
#fit = calculate mean + standard deviation
```

```python
print(scaled_data)
```
