```python
import pandas as pd
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.DataFrame({
    "Satisfaction": ["Low", "High", "Medium", "Low", "High"]
})

print(df)
```

```python
encoder = OrdinalEncoder(
   categories=[["Low", "Medium", "High"]]
)
```

```python
encoded = encoder.fit_transform(df[["Satisfaction"]])
```

```python
print(encoded)
```
